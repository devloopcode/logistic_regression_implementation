```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import loguniform

from sklearn.datasets import make_classification
from sklearn.pipeline import Pipeline
from sklearn.model_selection import RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
import os
os.environ['MallocStackLogging'] = '0'
```

```python
# Create the dataset
X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_classes=2,
    random_state=42    
)
```

```python
pd.DataFrame(X)
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    random_state=42,
    test_size=0.3
)
```

```python
logistic = LogisticRegression()
```

```python
logistic.fit(X_train, y_train)
```

```python
y_pred = logistic.predict(X_test)
```

```python
y_pred
```

```python
score = accuracy_score(y_test, y_pred)
print(f"score: {score}")
```

```python
cm = confusion_matrix(y_test, y_pred)
print("confusion matrix:")
print(cm)
```

```python
print(f"classification report:\n {classification_report(y_test, y_pred)}")
```

## Hyperparameter Tuning & Cross Validation

#### Grid SearchCV

```python
pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(max_iter=5000, random_state=42))
])

param_grid = [
    {
        'clf__solver': ['lbfgs', 'newton-cg', 'sag'],
        'clf__penalty': ['l2'],
        'clf__C': [0.01, 0.1, 1, 10],
    },
    {
        'clf__solver': ['lbfgs', 'newton-cg', 'sag'],
        'clf__penalty': [None],
    },
]
```

```python
# GridSearchCV
cv = StratifiedKFold()
grid = GridSearchCV(pipe, param_grid, scoring='accuracy', cv=5, n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred_grid = grid.predict(X_test)
```

```python
y_pred_grid
```

```python
score = accuracy_score(y_test, y_pred_grid)
print(f"score: {score}")
```

```python
cm = confusion_matrix(y_test, y_pred_grid)
print("confusion matrix:")
print(cm)
```

```python
print(f"classification report:\n {classification_report(y_test, y_pred_grid)}")
```

#### Randomized SearchCV

```python
random_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(max_iter=5000, random_state=42))
])

param_distributions = [
    {
        'clf__solver': ['lbfgs', 'newton-cg', 'sag'],
        'clf__penalty': ['l2'],
        'clf__C': loguniform(1e-3, 1e2),
    },
    {
        'clf__solver': ['lbfgs', 'newton-cg', 'sag'],
        'clf__penalty': [None],
    },
    {
        'clf__solver': ['liblinear'],
        'clf__penalty': ['l1', 'l2'],
        'clf__C': loguniform(1e-3, 1e2),
    },
    {
        'clf__solver': ['saga'],
        'clf__penalty': ['l1', 'l2'],
        'clf__C': loguniform(1e-3, 1e2),
    },
    {
        'clf__solver': ['saga'],
        'clf__penalty': [None],
    },
    {
        'clf__solver': ['saga'],
        'clf__penalty': ['elasticnet'],
        'clf__C': loguniform(1e-3, 1e2),
        'clf__l1_ratio': loguniform(0.01, 1),
    },
]
```

```python
random_search = RandomizedSearchCV(
    random_pipe,
    param_distributions=param_distributions,
    n_iter=50,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    random_state=42,
)
```

```python
random_search.fit(X_train, y_train)
```

```python
print("Best params:", random_search.best_params_)
print("Best CV score:", random_search.best_score_)
```

```python
y_pred_random_cv = random_search.predict(X_test)
```

```python
y_pred_random_cv
```

```python
score = accuracy_score(y_test, y_pred_random_cv)
print(f"score: {score}")
```

```python
cm = confusion_matrix(y_test, y_pred_random_cv)
print("confusion matrix:")
print(cm)
```

```python
print(f"classification report:\n {classification_report(y_test, y_pred_random_cv)}")
```
